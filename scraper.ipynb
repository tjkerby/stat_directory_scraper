```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
r = requests.get("https://statistics.byu.edu/faculty-directory")
soup = BeautifulSoup(r.text)
```

```python
raw_links = soup.find_all("a")
```

```python
links = [link['href'] for link in raw_links]
```

```python
print(len(links))
links = list(set(links))
print(len(links))
```

```python
final_links = [link for link in links if link.__contains__('statistics.byu.edu/directory')]
```

```python
len(final_links)
```

```python
data_dict = {
    "name": [], 
    "title": [], 
    'website': [], 
    'phone': [], 
    'email': [], 
    'office': []
}

def safe_scrape(soup, tag, class_name):
    try:
        return soup.find(tag, {"class": class_name}).text.strip()
    except:
        return None

for link in final_links:
    print(link)
    directory = requests.get(link)
    soup = BeautifulSoup(directory.text)

    name = safe_scrape(soup, "h1", "EmployeePage-name")
    title = safe_scrape(soup, "div", "EmployeePage-jobTitle")
    website = safe_scrape(soup, "div", "EmployeePage-websiteLink")
    phone = safe_scrape(soup, "div", "EmployeePage-phoneNumber")
    email = safe_scrape(soup, "div", "EmployeePage-email")
    office = safe_scrape(soup, "div", "EmployeePage-address")

    data_dict['name'].append(name)
    data_dict['title'].append(title)
    data_dict['website'].append(website)
    data_dict['phone'].append(phone)
    data_dict['email'].append(email)
    data_dict['office'].append(office)

df = pd.DataFrame(data_dict)
df.head()
```

```python
import re

def clean_website(website):
    pattern = r"https?://\w+\.byu\.edu"
    try:
        results = re.search(pattern, website)
        return results[0]
    except:
        return None
    
df["cleaned_website"] = df.loc[:, "website"].apply(clean_website)
df.head()
```

```python
def clean_office(office):
    pattern = r"\d{4}[A-Z]? [A-Z]{2,5}"
    try:
        results = re.search(pattern, office)
        return results[0]
    except:
        return None
    
df['cleaned_office'] = df.loc[:, "office"].apply(clean_office)
df.head()
```
